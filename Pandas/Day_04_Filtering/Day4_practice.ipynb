```python
import pandas as pd

data = {
    "Name": ["Amit", "Sneha", "Ravi", "Priya", "Kiran", "Anjali"],
    "Age": [25, 30, 22, 28, 35, 24],
    "City": ["Bangalore", "Mumbai", "Delhi", "Bangalore", "Mumbai", "Delhi"],
    "Salary": [50000, 65000, 40000, 55000, 75000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# df["Age"]>25

df[df["Age"]>25]
```

```python
# df["City"] == 'Bangalore'

df[df["City"] == "Bangalore"]
```

```python
df[df["Salary"]>=55000]
```

```python
df[
    (df["Age"] > 25) &
    (df["Salary"] > 50000)
]
```

```python
df[
    (df["City"] == "Delhi") |
    (df["Salary"] > 70000)
]
```
